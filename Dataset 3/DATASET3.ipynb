---
### 1. Renomeie as três variáveis de consumo para Consumo_Zona_1, Consumo_Zona_2 e Consumo_Zona_3.

```python
df.columns
```

```python
df.rename(columns={'Zone 1 Power Consumption':  'Consumo_Zona_1',
                   'Zone 2  Power Consumption': 'Consumo_Zona_2',
                   'Zone 3  Power Consumption': 'Consumo_Zona_3',
                   }, inplace=True)

df.head()
```

---
### 2. Determine o consumo máximo registrado em cada uma das três zonas.

```python
zonas = ['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']

max_consumo = df[zonas].max()
print(max_consumo)
```

---
### 3. Identifique qual zona apresenta o maior pico de consumo na amostra.

```python
zona_pico  = max_consumo.idxmax()
valor_pico = max_consumo.max()

print(f'A zona com o maior pico de consumo e: {zona_pico}')
print(f'Valor do pico: {valor_pico:.2f} kW')
```

---
### 4. Para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar.

```python
limiar  = valor_pico * 0.70
df_alto = df[df[zona_pico] > limiar]

print(f'Limiar (70% de {valor_pico:.2f}): {limiar:.2f} kW')
print()
df_alto.head()
```

---
### 5. Conte os registros e calcule o percentual que eles representam.

```python
qtd = len(df_alto)
pct = qtd / len(df) * 100

print(f'Quantidade de registros acima do limiar: {qtd}')
print(f'Percentual da amostra: {pct:.2f}%')
```

---
### 6. Calcule a temperatura média e crie um segundo DataFrame contendo consumo acima do limiar e temperatura acima da média.

```python
media_temp     = df['Temperature'].mean()
df_alto_quente = df_alto[df_alto['Temperature'] > media_temp]

qtd2 = len(df_alto_quente)
pct2 = qtd2 / len(df) * 100

print(f'Temperatura media da amostra: {media_temp:.2f} C')
print(f'Registros com consumo alto E temperatura acima da media: {qtd2}')
print(f'Percentual da amostra: {pct2:.2f}%')
```

```python
df_alto_quente.head()
```

---
### 7. Compare os dois conjuntos e explique o que ocorreu com a quantidade de registros após a inclusão da condição ambiental.

```python
reducao   = qtd - qtd2
retencao  = qtd2 / qtd * 100
base      = (df['Temperature'] > media_temp).mean() * 100

print(f'Conjunto 1 - consumo alto {qtd} registros ({pct:.2f}% da amostra)')
print(f'Conjunto 2 - consumo alto + calor {qtd2} registros ({pct2:.2f}% da amostra)')
print()
print(f'Registros removidos pela condicao ambiental: {reducao}')
print(f'Percentual de picos que ocorrem em alta temperatura: {retencao:.2f}%')
print(f'Percentual da amostra toda com temperatura acima da media: {base:.2f}%')
```

**Resposta:**

**O que ocorreu:** a quantidade caiu de **2.388 para 1.597** registros 791 a menos. A queda era esperada, porque o segundo critério é aplicado com o conector e dentro do primeiro conjunto: o conjunto 2 é necessariamente um subconjunto do conjunto 1 e só pode encolher.

**O que o resultado mostra:** o número relevante não é a queda, e sim a proporção que sobrou. Na amostra inteira, **49,51%** dos registros têm temperatura acima da média praticamente metade, o que é esperado por definição de média. Já entre os registros de alto consumo, esse percentual sobe para **66,88%**.

Ou seja, os picos de demanda da Zona 1 se concentram nos momentos mais quentes, o que aponta a carga térmica (refrigeração) como componente relevante da ponta. Mas o calor não explica todos eles: **33,12%** dos picos ocorrem com temperatura abaixo da média. A temperatura é um fator associado à alta demanda, não a sua única causa.
