#Situação

Uma equipe de operação de uma usina fotovoltaica deseja localizar períodos de alta geração e identificar quais inversores aparecem com maior frequência nesses momentos. Nesta atividade, o foco inicial será o arquivo de geração da planta, sem realizar junção com o arquivo de sensores.

#Etapa B — Python / Pandas

1. Carregue a amostra e renomeie as principais variáveis para nomes mais claros, como Potencia_CC, Potencia_CA e Geracao_Diaria.

```python
import pandas as pd

df1 = pd.read_csv('/content/AMOSTRA-DataSet-CP03.csv')
df1.head()
```

```python
df1.rename(columns={
    'DATE_TIME': 'Data_Hora',
    'SOURCE_KEY': 'Chave_Origem',
    'DC_POWER': 'Potencia_CC',
    'AC_POWER': 'Potencia_CA',
    'DAILY_YIELD': 'Geracao_Diaria',
    'TOTAL_YIELD': 'Geracao_Total'
}, inplace=True)
```

#3. Determine a maior potência CA registrada.

```python
PMAX = df1['Potencia_CA'].max()
print(f'O valor máximo de demanda energética é: {PMAX:.2f} kW')
```

#4. Calcule um limiar correspondente a 70% da potência CA máxima.

```python
P70 = 0.7 * PMAX
print(f'O limiar de 70% da demanda máxima é: {P70:.2f} kW')
```

#5. Crie um DataFrame contendo somente registros acima desse limite e calcule sua quantidade e percentual.

```python
df2 = df1[df1['Potencia_CA'] > P70]
print(f'Total de registros acima do limiar: {df2.shape[0]}')
```

```python
percentual_acima_limiar = (df2.shape[0] / df1.shape[0]) * 100
print(f'Percentual de registros acima do limiar: {percentual_acima_limiar:.2f}%')
```

#6. Utilize value_counts() em SOURCE_KEY dentro do DataFrame de alta geração.

```python
contagem_inversores = df2['Chave_Origem'].value_counts()
print(f"\n{contagem_inversores}")
```

#7. Identifique qual inversor aparece com maior frequência nos registros de alta geração e descreva o que esse resultado permite observar, sem concluir desempenho ou falha apenas com esse recorte.

```python
inversor_mais_frequente = contagem_inversores.idxmax()
quantidade_maxima = contagem_inversores.max()
print(f'O inversor mais frequente é: {inversor_mais_frequente}')
print(f'Quantidade de ocorrências: {quantidade_maxima}')
```

**Observação: Este resultado indica quais equipamentos estiveram operando acima de 70% da capacidade máxima
mais vezes nesta amostra. Isso permite observar o comportamento operacional da planta no período,
mas não é suficiente para concluir sobre o desempenho superior ou falhas, pois outros fatores
(como localização física ou irradiação local) não foram analisados. bold text**
